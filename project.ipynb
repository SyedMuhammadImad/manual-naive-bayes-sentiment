## Scratch implementation check

A mathematical fixture verifies procedural word counts and log inference; it is not measured corpus performance.


```python
texts=['good excellent','bad terrible','ordinary standard']
labels=np.array(['positive','negative','neutral'])
vocab=fit_vocabulary(texts)
model=train_naive_bayes(bag_of_words(texts,vocab),labels)
assert custom_inference('good excellent',model,vocab)[0]=='positive'
assert np.isfinite(predict_log_proba(bag_of_words(['unknown'],vocab),model)).all()
```
